Q3. Create orders.csv, load it into a DataFrame and perform set-based profiling: (a) which columns
contain categorical data? (b) find unique values of each categorical variable and count the frequency
of each category, (c) list the most frequent regions.

```python
import pandas as pd
df = pd.read_csv("order.csv")
categorical_cols = ['Customer', 'Region', 'Product']
print("(a) Categorical columns:", categorical_cols)
print("(b) Customer")
print("Unique values:", df['Customer'].unique().tolist())
print(df['Customer'].value_counts())

print("(b) Region")
print("Unique values:", df['Region'].unique().tolist())
print(df['Region'].value_counts())

print("(b) Product")
print("Unique values:", df['Product'].unique().tolist())
print(df['Product'].value_counts())

most_frequent_region = df['Region'].mode().tolist()
count = df['Region'].value_counts().max()

print("(c) Most frequent region(s):", most_frequent_region)
print("Count:", count)
```
