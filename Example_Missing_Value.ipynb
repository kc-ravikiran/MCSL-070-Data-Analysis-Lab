**Hospital ICU Monitoring Data**
This data comes from continuous monitoring equipment in a high-intensity environment
like an ICU, where hardware failures are common.

The first step in handling missing data is to identify where the gaps exist. The pandas
library provides intuitive functions for this:

Let us start by identifying missing values using pandas:

```python
# Importing necessary libraries
import pandas as pd
import numpy as np
```

```python
# Create a sample dataset with missing values
icu_data = {
'admission_id': ['I201', 'I201', 'I202', 'I201', 'I202', 'I202'],
'minute_timestamp': [
"2025-07-06 10:00",
"2025-07-06 10:03",
"2025-07-06 10:01",
"2025-07-06 10:07",
"2025-07-06 10:05",
"2025-07-06 10:09"
],
'heart_rate': [85, 87, 75, np.nan, 78, 79],
# Missing due to detached ECG lead (I201, T+3min)
'systolic_bp': [120, 118, np.nan, 122, 125, 124],
# Missing due to patient movement/skipped cuff cycle (I202, T+2min)
'respiratory_rate': [18, 18, 16, 17, np.nan, 16]
# Missing due to temporary ventilator disconnect (I202, T+4min)
}
df = pd.DataFrame(icu_data)
```

```python
print("\n--- Hospital ICU Monitoring Data (EquipmentFailure) ---")
print(df)
```

The following segment detects the missing value from the program above.

```python
# Detect Missing Values
# Identifying missing values using.isna() or.isnull()
# Both functions work the same way and return a Boolean DataFrame
missing_values_df = df.isna()
print("\nDataFrame showing missing values (True for NaN):")
print(missing_values_df)
```

The panda function isna() is used to find out the missing values in the Dataframe. The
function creates a Dataframe containing only Boolean data.

```python
# A more practical approach is to count missing values per column
missing_counts = df.isnull().sum()
print("\nTotal missing values per column:")
print(missing_counts)
```

```python
# We can also calculate the percentage of missing values
total_rows = len(df)
percentage_missing = (missing_counts / total_rows) * 100
print("\nPercentage of missing values per column:")
print(percentage_missing)
```
