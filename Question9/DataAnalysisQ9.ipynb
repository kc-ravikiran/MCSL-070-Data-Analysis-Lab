Q9.Create a dataset named employee_attendance.csv that contains the
following records.
Use this data set to perform the following tasks relating to missing
values from the dataset as per the following tasks using Python code:
a) List those rows that contain missing value in any column.
b) Clean the dataset so that there is no row with any missing
value.
c) Find number of rows which were removed during cleaning.
d) Display the cleaned dataset.

```python
import pandas as pd

# Load dataset
df = pd.read_csv("employee_attendance.csv")

print("Original Dataset:")
print(df)

# (a) Rows with missing values
missing_rows = df[df.isnull().any(axis=1)]

print("\n(a) Rows containing missing values:")
print(missing_rows)

# (b) Remove rows with missing values
cleaned_df = df.dropna()

# (c) Number of rows removed
rows_removed = len(df) - len(cleaned_df)

print("\n(c) Number of rows removed:", rows_removed)

# (d) Display cleaned dataset
print("\n(d) Cleaned Dataset:")
print(cleaned_df)
```
