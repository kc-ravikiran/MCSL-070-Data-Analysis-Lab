Q10. A dataset Student_Marks.csv is given below. Use this data set to
perform tasks relating to replacement of Missing Values using
Mean/Median/Mode as asked..
a) Replace missing marks in Maths with the mean score in Maths
examination.
b) Replace missing marks in Science with the median score in
Science
c) Replace missing marks in English with the mode score in
English examination.
d) Display the modified dataset highlighting the values that are
# used fro replacing the missing values.

```python
import pandas as pd
df = pd.read_csv("Student_Marks.csv")
print(df)
```

```python
# Calculate replacement values
maths_mean = df["Marks in Maths"].mean()
science_median = df["Marks in Science"].median()
english_mode = df["Marks in English"].mode()[0]

# Replace missing values
df["Marks in Maths"] = df["Marks in Maths"].fillna(maths_mean)
df["Marks in Science"] = df["Marks in Science"].fillna(science_median)
df["Marks in English"] = df["Marks in English"].fillna(english_mode)

# Display modified dataset
print("Modified Dataset:")
print(df)

# Display replacement values
print("\nReplaced Values:")
print("Roll No 1002 -> Maths =", maths_mean)
print("Roll No 1003 -> Science =", science_median)
print("Roll No 1004 -> English =", english_mode)
```
