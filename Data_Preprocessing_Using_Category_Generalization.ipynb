Q. Create a Pandas DataFrame from the given consumer electronics dataset and perform category generalization using a mapping technique.

```python
import pandas as pd
# Dataset contains List of Dictionaries
ConsumerElectronics = [
{"ConsumerName": "Manish", "Model": "Reno",
"ProductType": "Smartphone"},
{"ConsumerName": "Meenu", "Model": "Thinkpad",
"ProductType": "Laptop"},
{"ConsumerName": "Rafiq", "Model": "Galaxy",
"ProductType": "Tablet"},
{"ConsumerName": "Samuel", "Model": "MacBookAir",
"ProductType": "Laptop"},
{"ConsumerName": "Vikrant", "Model": "Edge60",
"ProductType": "Smartphone"},
{"ConsumerName": "Kiara", "Model": "Pad3",
"ProductType": "Tablet"},
{"ConsumerName":"Kiran", "Model": "Oppo F29", "ProductType": "Smart Phone"}
]
# Create DataFrame
Consumerdf = pd.DataFrame(ConsumerElectronics)
print("Consumer DataFrame:")
print(Consumerdf)
# Use of an "Is-a" Mapping
category_map = {
"SimpleMobile": "Mobiles",
"Smartphone": "Mobiles",
"Laptop": "Portable Computing",
"Tablet": "Portable Computing"
}
# New column created from generalisation of category
Consumerdf["New_Category"] = Consumerdf["ProductType"].map(category_map)
print("DataFrame with New Category:")
print(Consumerdf)
```
