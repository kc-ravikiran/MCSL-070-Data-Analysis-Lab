A dataset Sales_Records.csv is given below. Rename the columns of
this data set as the following task:
Refer CSV file
a) Rename the column Order_ID to Order; Total Sales to
Sales; and Profit/loss to Profit.
b) Display the updated dataset with modified column names.

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("Sales_Records.csv")
print("Original Data")
print(df)

# (a) Rename columns
df.rename(columns={
    "Order_ID": "Order",
    "Total Sales": "Sales",
    "Profit/loss": "Profit"
}, inplace=True)

# (b) Display updated dataset
print("Updated Dataset:")
print(df)
```
