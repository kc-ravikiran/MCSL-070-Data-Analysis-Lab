A dataset Sales_Records.csv is given below. Filter the data with nonnull
values in specific columns, as asked in the following tasks:
Refer Sales_Records.csv

a) Filter only those rows where both Total Sales and Profit/loss
columns have non-null values.
b) Display the filtered dataset.
c) Find out how many rows satisfy this condition.

```python
import pandas as pd

df = pd.read_csv("Sales_Records.csv")
```

```python
# (a) Filter rows where both Total Sales and Profit/loss are non-null
filtered_df = df[df["Total Sales"].notna() & df["Profit/loss"].notna()]

# (b) Display the filtered dataset
print("Filtered Dataset:")
print(filtered_df)

# (c) Count the number of rows satisfying the condition
count_rows = len(filtered_df)

print("\nNumber of rows with non-null values in both columns:", count_rows)
```
