Q5. Create salary.csv, load it into a DataFrame and perform distribution analysis: (a) find mean,
median and mode of salaries, (b) is there any skewness in the data? Give reasons, (c) use
descriptive statistics to find the outliers.

```python
import pandas as pd
from scipy.stats import skew

# Load CSV file
df = pd.read_csv("salary.csv")

# (a) Mean, Median, Mode
print("(a) Mean :", df["Salary"].mean())
print("Median:", df["Salary"].median())
print("Mode :", df["Salary"].mode().tolist())

# (b) Skewness
print("\n(b) Skewness:", round(skew(df["Salary"]), 3))

# (c) Descriptive Statistics
print("\n(c) Descriptive statistics:")
print(df["Salary"].describe())

# Quartiles and IQR
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)
IQR = Q3 - Q1

print("\nQ1 =", Q1, "Q3 =", Q3, "IQR =", IQR)

# Lower and Upper fences
lower_fence = Q1 - (1.5 * IQR)
upper_fence = Q3 + (1.5 * IQR)

print("Lower fence =", lower_fence, "Upper fence =", upper_fence)

# Detect Outliers
outliers = df[(df["Salary"] < lower_fence) | (df["Salary"] > upper_fence)]

print("Outliers:")
print(outliers)
```
