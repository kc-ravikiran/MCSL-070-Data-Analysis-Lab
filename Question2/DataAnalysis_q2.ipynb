Q2. Create employees.csv, load it into a DataFrame and: (a) find the datatype of each column, (b)
convert Joining_Date to datetime and verify, (c) find mean, median, variance and standard deviation
of Salary, (d) assuming Salary is annual, find the total salary paid to all employees.

```python
import pandas as pd
df = pd.read_csv("employees.csv")
print("Datatypes of Columns:")
print(df.dtypes)
```

```python
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])

print(df.dtypes)
```

```python
print("Mean Salary:", df["Salary"].mean())
```

```python
print("Median Salary:", df["Salary"].median())
```

```python
print("Variance:", df["Salary"].var())
```

```python
print("Standard Deviation:", df["Salary"].std())
```

```python
total_salary = df["Salary"].sum()
print("Total Annual Salary Paid:", total_salary)
```

```python
import pandas as pd

# Load CSV file
df = pd.read_csv("employees.csv")

# Display datatypes
print("Datatypes of Columns:")
print(df.dtypes)

# Convert Joining_Date to datetime
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])

# Verify conversion
print("\nDatatypes After Conversion:")
print(df.dtypes)

# Salary statistics
print("\nMean Salary:", df["Salary"].mean())
print("Median Salary:", df["Salary"].median())
print("Variance:", df["Salary"].var())
print("Standard Deviation:", df["Salary"].std())

# Total annual salary paid
print("\nTotal Annual Salary Paid:", df["Salary"].sum())
```
