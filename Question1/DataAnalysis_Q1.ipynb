```python
import pandas as pd

# Load CSV file
df = pd.read_csv("studentdata.csv")

# Display first 2 rows
print("First 2 Rows:")
print(df.head(2))

# Display last 3 rows
print("\nLast 3 Rows:")
print(df.tail(3))

# Display dimensions
print("\nDimensions:")
print(df.shape)

# Display column names
print("\nColumn Names:")
print(df.columns)
```
