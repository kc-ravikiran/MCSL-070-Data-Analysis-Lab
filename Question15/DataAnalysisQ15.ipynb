Q15.Create a dataset named employee_attendance.csv that contains the
following records. Replace the abbreviations in the Department using
the following and display the modified data set.
HR to Human Resources
IT to Information Technology

```python
import pandas as pd

# Load the CSV file
df = pd.read_csv("employee_attendance.csv")

print("Original Data")
print(df)

# Replace department abbreviations
df["Department"] = df["Department"].replace({
    "HR": "Human Resources",
    "IT": "Information Technology"
})

# Display modified dataset
print(df)
```

```python
#Adding new row
df.loc[len(df)] = [1006, "Kiran", "CEO", 0]

# Display updated dataset
print(df)
```

```python
# Adding one more
df.loc[len(df)] = [1007, "Theju", "MG", 0]

print(df)
```
