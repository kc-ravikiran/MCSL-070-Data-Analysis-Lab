Q8. Create answers.csv, load it into a DataFrame and deal with correlation and redundancy: (a)
compute the correlation matrix, (b) list the highly correlated columns having a correlation close to 1.
Find a group of students with almost similar answers. Give reasons.

```python
import pandas as pd
```

```python
df = pd.read_csv("answers.csv")

print(df)
```

```python
corr_matrix = df.corr()

print("(a) Correlation matrix:")
print(corr_matrix)
```

```python
threshold = 0.9

print("\n(b) Highly correlated pairs (correlation > 0.9):")

for i in range(len(corr_matrix.columns)):
    for j in range(i + 1, len(corr_matrix.columns)):
        corr_value = corr_matrix.iloc[i, j]

        if abs(corr_value) > threshold:
            print(
                corr_matrix.columns[i],
                "&",
                corr_matrix.columns[j],
                "->",
                round(corr_value, 2)
            )
```
