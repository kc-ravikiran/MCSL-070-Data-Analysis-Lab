Q6. Create city_data.csv, load it into a DataFrame and find inconsistent data: (a) find all the
inconsistent entries, (b) standardise and convert all city names to uppercase, (c) count the names of
unique cities.

```python
import pandas as pd
df = pd.read_csv("city_data.csv")
```

```python
# (a) Entries as stored
print("(a) Entries as stored:")
print(df["City"].value_counts())
```

```python
# Count distinct spellings
print("\nDistinct spellings:", df["City"].nunique())
```

```python
# Find inconsistent entries
valid_entries = ["Delhi", "Mumbai"]
inconsistent = df[~df["City"].isin(valid_entries)]
print("\nInconsistent entries (not in the form Delhi / Mumbai):")
print(inconsistent)
```

```python
# (b) Convert city names to uppercase
df["City"] = df["City"].str.upper()
print("\n(b) After converting to uppercase:")
print(df)
```

```python
# (c) Count unique cities after standardization
print("\n(c) Number of unique cities:", df["City"].nunique())
print(df["City"].value_counts())
```
