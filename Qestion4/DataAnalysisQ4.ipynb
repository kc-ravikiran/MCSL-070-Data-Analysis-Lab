Q4. Create marks.csv, load it into a DataFrame and perform missing value analysis: (a) identify
missing values in each column, (b) compute the percentage of missing values per column, (c)
identify the row with the maximum missing values.

```python
import pandas as pd

# Load CSV file
df = pd.read_csv("marks.csv")

# (a) Missing values per column
print("(a) Missing values per column:")
print(df.isnull().sum())

# (b) Percentage of missing values per column
print("\n(b) Percentage missing per column:")
print((df.isnull().sum() / len(df)) * 100)

# (c) Missing values per row
df["missing"] = df.isnull().sum(axis=1)

print("\n(c) Missing values per row:")
print(df["missing"])

# Row(s) with maximum missing values
max_missing = df["missing"].max()

print("\nRow(s) with the maximum missing values:")
print(df[df["missing"] == max_missing].drop(columns=["missing"]))
```
