Q11: A dataset Customer_Info.csv contains missing categorical data. Fill
this Missing Categorical data as per the following tasks:a) Find the missing values in the column named Membership.
b) Replace the missing values with the placeholder "Not
Shared" and display the dataset
c) Count how many missing values were replaced in the dataset.

```python
import pandas as pd

df = pd.read_csv("Customer_Info.csv")

print(df)
```

```python
# Find missing values
missing_count = df["Membership"].isnull().sum()
print("Missing values in Membership column:", missing_count)

# Replace missing values
df["Membership"] = df["Membership"].fillna("Not Shared")

# Display updated dataset
print("\nUpdated Dataset:")
print(df)

# Count replaced values
replaced_count = (df["Membership"] == "Not Shared").sum()
print("\nNumber of missing values replaced:", replaced_count)
```
