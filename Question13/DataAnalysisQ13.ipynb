A dataset Orders.csv contains duplicate rows. Perform the following
tasks for this data set.
refer Records.csv
a) Find the duplicate rows in the data set.
b) Remove the duplicate rows. Just keep the first occurrence of a
duplicate row in the data.
c) Display the cleaned dataset and identify the number of duplicate
rows removed.

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("Orders.csv")
print("Original Data")
print(df)

# (a) Find duplicate rows
duplicates = df[df.duplicated()]

print("Duplicate Rows:")
print(duplicates)

# (b) Remove duplicate rows and keep first occurrence
cleaned_df = df.drop_duplicates(keep='first')

# (c) Display cleaned dataset
print("\nCleaned Dataset:")
print(cleaned_df)

# Number of duplicate rows removed
removed_rows = len(df) - len(cleaned_df)

print("\nNumber of duplicate rows removed:", removed_rows)
```
