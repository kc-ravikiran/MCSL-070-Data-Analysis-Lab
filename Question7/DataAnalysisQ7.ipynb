Q7. Create employeeSet1.csv and employeeSet2.csv, load them into two DataFrames and compare
them: (a) find the rows common to Dataset 1 and Dataset 2, (b) find the rows present in Dataset 1
but not in Dataset 2, and the rows present in Dataset 2 but not in Dataset 1.

```python
import pandas as pd
df1 = pd.read_csv("employeeset1.csv")
df2 = pd.read_csv("employeeset2.csv")

print("Dataset 1:")
print(df1)

print("\nDataset 2:")
print(df2)
```

```python
common_rows = pd.merge(df1, df2, how='inner')

print("(a) Rows common to both:")
print(common_rows)
```

```python
only_df1 = df1.merge(df2, how='left', indicator=True)
only_df1 = only_df1[only_df1['_merge'] == 'left_only']
only_df1 = only_df1.drop('_merge', axis=1)

print("\n(b) In Dataset 1 but not in Dataset 2:")
print(only_df1)
```

```python
only_df2 = df2.merge(df1, how='left', indicator=True)
only_df2 = only_df2[only_df2['_merge'] == 'left_only']
only_df2 = only_df2.drop('_merge', axis=1)

print("\nIn Dataset 2 but not in Dataset 1:")
print(only_df2)
```
